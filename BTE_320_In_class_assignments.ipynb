Assignment 1

```python
sales= float(input("Give the sales value($):"))
tax_rate= float(input("Give the tax rate(%):"))
print (f"Sales: {sales}$")
COGS = sales*.4
print (f"Cost of Goods Sold: {COGS}$")
profit= sales-COGS
print (f"Gross Profit: {profit}$")
taxes= profit*tax_rate/100
print(f"Taxes: {taxes}$")
net= profit-taxes
print(f"Net Income: {net}$")
```

```python
num_1= float(input("Give the first number"))
num_2= float(input("Give the second number"))
operation= (input("Give the operation"))

if operation == "+":
  print (num_1 + num_2)

elif operation == "-":
  print(num_1 - num_2)

elif operation =="*":
  print (num_1 * num_2)

else:
  print("invalid operator. please use +,-,*")
```
