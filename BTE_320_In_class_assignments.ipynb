```python
class Greeting:
  def __init__(self, first,last):
    self.first= first
    self.last=last

  def display (self, formal=False):
    if formal:
      print(f"{self.last}, {self.first}")
    else:
      print(self.first)

  def __str__(self):
    return "Instance of class 'Greeting'"
name=Greeting("John","Doe")
print(name)
name.display()
name.display(formal=True)
```

```python
from os import name
class Student:
  def __init__(self,name,number,courses):
    self.__name=name
    self.number=number
    self.__courses=courses

  def get_name(self):
    return self.__name

  def set_name(self,newname):
    self.__name=newname

  def get_courses(self):
    return self.__courses

  def __len__(self):
    return len(self.__courses)

  def enroll(self,newCourse):
    if newCourse not in self.__courses:
      self.__courses.append(newCourse)
    else:
      print(f"Student {self.__name} is already enrolled in {newCourse}.")
```

```python
s=Student("Jane Doe", 12345, [])
print(f"Student name: {s.get_name()}")
print(f"Classes already enrolled in: {s.get_courses()}")
s.enroll('BTE 320')
print(f"Classes already enrolled in: {s.get_courses()}")
s.enroll('BTE 320')
print (f"Total number of classes enrolled: {len(s)}")
```
